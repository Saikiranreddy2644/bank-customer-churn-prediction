```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

df = pd.read_csv("data/raw/churn_dirty.csv")
print(f"Shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
```

```python
missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)

missing_report = pd.DataFrame({
    "Missing Count": missing,
    "Missing %": missing_pct
}).sort_values("Missing Count", ascending=False)

print(missing_report[missing_report["Missing Count"] > 0])
```

```python
print(f"Exact duplicate rows : {df.duplicated().sum()}")
print(f"Duplicate CustomerIds: {df.duplicated(subset=['CustomerId']).sum()}")
```

```python
print(f"Geography unique values: {df['Geography'].unique()}")
print(f"Gender unique values   : {df['Gender'].unique()}")
```

```python
print(f"Age min: {df['Age'].min()}  Age max: {df['Age'].max()}")
print(f"Impossible ages (< 18 or > 100)     : {df[(df['Age'] < 18) | (df['Age'] > 100)].shape[0]} rows")

print(f"\nCreditScore min: {df['CreditScore'].min()}  max: {df['CreditScore'].max()}")
print(f"Invalid CreditScores (< 300 or > 850): {df[(df['CreditScore'] < 300) | (df['CreditScore'] > 850)].shape[0]} rows")

print(f"\nNegative salaries: {df[df['EstimatedSalary'] < 0].shape[0]} rows")
```

```python
print(df["Exited"].value_counts())
print(f"\nChurn rate: {round(df['Exited'].mean() * 100, 2)}%")
```
